```python
# --- SISTEMA DE IA: AGENTE DE CASTING "TALENT-SCAN" ---

# Inicialización: La lista de actores empieza vacía
lista_actores = []

def registrar_actor():
    print("\n--- REGISTRO DE NUEVO TALENTO ---")
    # Se crea un diccionario vacío para el actor actual
    actor = {}

    # Entrada de datos (Key: Value)
    nombre = input("Nombre del Actor/Actriz: ")
    actor["Nombre"] = nombre
    actor["Agencia"] = input(f"Agencia de representación de {nombre}: ")
    actor["Obras"] = int(input(f"Cantidad de obras realizadas por {nombre}: "))
    actor["Premios"] = int(input("Cantidad de premios obtenidos: "))
    actor["Habilidades"] = input("Habilidades especiales (ej. artes marciales, canto, baile): ").lower()
    actor["RangoEdad"] = int(input("Rango de edad (años): "))
    actor["Proyecto"] = input("Nombre del último proyecto: ")

    # Se agrega el diccionario a la lista global
    lista_actores.append(actor)
    print(f"\n¡{nombre} ha sido integrado a la base de datos!")

def listar_actores():
    if len(lista_actores) == 0:
        print("\n[!] La base de datos está vacía actualmente.")
    else:
        print("\n" + "="*115)
        # Encabezado de la tabla con alineación fija
        print(f"{'NOMBRE':<15} | {'AGENCIA':<15} | {'OBRAS':<8} | {'PREMIOS':<8} | {'EDAD':<6} | {'HABILIDADES':<20} | {'ÚLTIMO PROYECTO':<15}")
        print("-" * 115)

        # Ciclo for para recorrer la lista de diccionarios
        for a in lista_actores:
            print(f"{a['Nombre']:<15} | {a['Agencia']:<15} | {a['Obras']:<8} | {a['Premios']:<8} | {a['RangoEdad']:<6} | {a['Habilidades']:<20} | {a['Proyecto']:<15}")
        print("="*115)

def ia_analizar_casting():
    if len(lista_actores) == 0:
        print("\n[IA] No hay perfiles para analizar. Por favor, registre actores primero.")
    else:
        print("\n[IA] Iniciando análisis de perfiles para recomendaciones de producción...")
        for a in lista_actores:
            print(f"\n> Analizando a: {a['Nombre']}...")

            # Motor de reglas de la IA utilizando if, elif, else y operadores lógicos
            if a['Premios'] >= 2 and a['Obras'] > 10:
                diagnostico = "PERFIL: 'ESTRELLA PROTAGONISTA'. Recomendado para roles principales en Drama/Cine de Culto."
            elif "artes marciales" in a['Habilidades'] or "acrobacia" in a['Habilidades']:
                diagnostico = "PERFIL: 'ACCIÓN/STUNT'. Ideal para películas de alto riesgo y superhéroes."
            elif a['RangoEdad'] < 20 and a['Obras'] <= 2:
                diagnostico = "PERFIL: 'TALENTO EMERGENTE'. Recomendado para series juveniles y nuevos lanzamientos."
            elif "canto" in a['Habilidades'] or "baile" in a['Habilidades']:
                diagnostico = "PERFIL: 'ARTISTA INTEGRAL'. Recomendado para Teatro Musical y producciones de Broadway."
            else:
                diagnostico = "PERFIL: 'ACTOR DE SOPORTE'. Versátil para roles secundarios y comerciales."

            print(diagnostico)

# --- MENÚ PRINCIPAL INTERACTIVO ---
while True:
    print("\n========================================")
    print("      SISTEMA IA: TALENT-SCAN 2.0      ")
    print("========================================")
    print("1. Registrar nuevo actor/actriz")
    print("2. Ver tabla de talentos")
    print("3. IA: Ejecutar análisis de Casting")
    print("4. Salir")

    opcion = input("\nSeleccione una opción (1-4): ")

    if opcion == "1":
        registrar_actor()
    elif opcion == "2":
        listar_actores()
    elif opcion == "3":
        ia_analizar_casting()
    elif opcion == "4":
        print("\nApagando Talent-Scan... ¡Mucha suerte en el set!")
        print("Programa realizado por Anais Medina 099 ")
        break
    else:
        print("\n[!] Opción no válida. Intente de nuevo.")

print("\n--- Sistema fuera de línea ---")
```
